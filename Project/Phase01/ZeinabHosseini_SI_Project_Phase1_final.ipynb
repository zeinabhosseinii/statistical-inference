**Negative values:**
Negative values in the `availability 365` column may not make sense, as the column represents the number of available days in a year. To handle this, we first calculate the median value of the column, considering only non-negative values. Then, we replace all negative values in the column with this calculated median value using the loc indexer.

**Replace Values Greater than 365 with 365:**
Similarly, values greater than 365 in the `availability 365` column may not be valid, as there are only 365 days in a year. We replace any values exceeding 365 with 365 using the loc indexer.

**Calculate Median and Replace Missing Values:**
After handling the negative and out-of-range values, we calculate the median value of the `availability 365` column. This median value is then used to replace any remaining missing values (NaN) in the column using the fillna method with inplace=True to modify the DataFrame directly.

```python
# Replace negative values with a suitable value (median)
median_value = df['availability 365'][df['availability 365'] >= 0].median()
df.loc[df['availability 365'] < 0, 'availability 365'] = median_value

# Replace values greater than 365 with a suitable value
df.loc[df['availability 365'] > 365, 'availability 365'] = 365

# Calculate the median for column 'availability 365'
median_value = df['availability 365'].median()

# Replace NaN values in column 'availability 365' with the median
df['availability 365'].fillna(median_value, inplace=True)
```

```python
df.info()
```

Filling null values in a DataFrame can be approached in various ways depending on the context and type of data in each column.

1. **For Categorical Columns (`object` dtype):**
    - **Mode:** We Replace nulls with the most common value in the column.

2. **For Numerical Columns (`float64` dtype):**
    - **Mean/Median:** We  Replace nulls with the mean or median of the column.
    
Median is less sensitive to outliers compared to the mean, making it a good choice for numerical data.

```python
# Filling null values for categorical columns
df['host_identity_verified'].fillna(df['host_identity_verified'].mode()[0], inplace=True)
df['neighbourhood group'].fillna(df['neighbourhood group'].mode()[0], inplace=True)
df['neighbourhood'].fillna(df['neighbourhood'].mode()[0], inplace=True)
df['instant_bookable'].fillna(df['instant_bookable'].mode()[0], inplace=True)
df['cancellation_policy'].fillna(df['cancellation_policy'].mode()[0], inplace=True)
df['price'].fillna(df['price'].mode()[0], inplace=True)
df['service fee'].fillna(df['service fee'].mode()[0], inplace=True)

# Filling null values for numerical columns
df['lat'].fillna(df['lat'].median(), inplace=True)
df['long'].fillna(df['long'].median(), inplace=True)
df['Construction year'].fillna(df['Construction year'].median(), inplace=True)
df['minimum nights'].fillna(df['minimum nights'].median(), inplace=True)
df['number of reviews'].fillna(df['number of reviews'].median(), inplace=True)
df['reviews per month'].fillna(df['reviews per month'].median(), inplace=True)
df['review rate number'].fillna(df['review rate number'].median(), inplace=True)
df['calculated host listings count'].fillna(df['calculated host listings count'].median(), inplace=True)

# Verify nulls are filled
df.isnull().sum()
```

**Categorize this variable into four intervals based on its mean and plot a pie chart that visualizes the frequency of these four categories. Your chart should be colorized, and the labels should contain each category with its percentage.**

```python
import pandas as pd
import matplotlib.pyplot as plt

temp = df.copy()

# Define intervals based on mean
mean_long = temp['long'].mean()
interval1 = mean_long - temp['long'].std()
interval2 = mean_long
interval3 = mean_long + temp['long'].std()
interval4 = mean_long + 2 * temp['long'].std()

# Categorize the 'long' variable into four intervals
def categorize_longitude(longitude):
    if longitude < interval1:
        return 'Interval 1'
    elif interval1 <= longitude < interval2:
        return 'Interval 2'
    elif interval2 <= longitude < interval3:
        return 'Interval 3'
    else:
        return 'Interval 4'

temp['category'] = temp['long'].apply(categorize_longitude)

# Calculate the frequency of each category
category_counts = temp['category'].value_counts()

# Plot a pie chart
plt.figure(figsize=(8, 8))
colors = ['lightblue', 'lightgreen', 'lightcoral', 'lightsalmon']
plt.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%', colors=colors, startangle=140)
plt.title('Frequency of Categories based on Longitude Intervals')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

## **C)**

**Calculate the correlation coefficient for these two variables. Using the “cor.test” function, we can also test the significance of a correlation. Are the variables correlated? According to the test, what is shown by the p-value, and what is the intuition of the p-value?**

```python
from scipy.stats import pearsonr

# Calculate Pearson correlation coefficient and p-value
correlation_coefficient, p_value = pearsonr(df['long'], df['lat'])

print("Correlation Coefficient:", correlation_coefficient)
print("P-value:", p_value)
```

1. **Correlation Coefficient**: The correlation coefficient measures the strength and direction of the linear relationship between two variables. In this case, the correlation coefficient between longitude ('long') and latitude ('lat') is positive, indicating a weak positive linear relationship between the variables.

2. **P-value**: The null hypothesis is that there is no correlation between longitude and latitude. The extremely small p-value (approximately 2.64e-125) indicates strong evidence against the null hypothesis. Therefore, we reject the null hypothesis and conclude that there is a statistically significant correlation between longitude and latitude.

While there is a statistically significant correlation between longitude and latitude due to the small p-value, the correlation coefficient suggests that the relationship between the two variables is very weak. This implies that changes in longitude are not strongly associated with changes in latitude, or vice versa.

## **B)**

**Design a hypothesis test to see if these data provide convincing
evidence of a difference between mean values. Does the result agree
with the 95% confidence interval?**

**Hypothesis**

- **Null Hypothesis $(H_0)$:** The means of the longitude and latitude are equal,  $\mu_{\text{long}} = \mu_{\text{lat}}$.
- **Alternative Hypothesis $(H_1)$:** The means of the longitude and latitude are not equal, $\mu_{\text{long}} \neq \mu_{\text{lat}}$.



```python
import numpy as np
from scipy import stats

# Perform two-sample t-test
t_stat, p_value = stats.ttest_ind(sample_df['lat'], sample_df['long'])

# Calculate the mean and standard deviation
mean_lat = np.mean(df['lat'])
mean_long = np.mean(df['long'])
std_lat = np.std(df['lat'], ddof=24)
std_long = np.std(df['long'], ddof=24)

# Sample sizes
n_lat = len(sample_df['lat'])
n_long = len(sample_df['long'])

# Standard error
se = np.sqrt(std_lat**2/n_lat + std_long**2/n_long)

# Margin of error at 95% confidence
z = stats.norm.ppf(0.975)  # 97.5th percentile value of the normal distribution
margin_of_error = z * se

# Confidence interval
ci_lower = (mean_lat - mean_long) - margin_of_error
ci_upper = (mean_lat - mean_long) + margin_of_error

# Results output
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")
print(f"95% Confidence Interval for the difference between means: ({ci_lower}, {ci_upper})")

# Decision based on p-value
alpha = 0.05  # Significance level
if p_value < alpha:
    print("Reject the null hypothesis - There is a significant difference between the means of the two variables.")
else:
    print("Fail to reject the null hypothesis - There is no significant difference between the means of the two variables.")
```

Based on the statistical analysis, the T-statistic is extremely high at
 $7964.8$, and the p-value is effectively zero  $(1.419e-148)$ , indicating a statistically significant difference between the mean values of latitude and longitude. The  $95$% confidence interval for the difference between these means ranges from  $114.648$  to  $114.707$ , which does not include zero. This results express to rejecte of the null hypothesis and also express that the means of the two variables are significantly different. Thus, the data provide strong evidence that the geographical coordinates represented by latitude and longitude have distinctly different average values in this dataset.

**Calculate a 95% confidence interval for the median of this variable using the percentile method and show the interval on the histogram.**

```python
from sklearn.utils import resample

# Set seed for reproducibility
np.random.seed(40)

# Sample 25 data points from 'number of reviews'
sample = df.sample(n=25, random_state=40)

# Select the feature from the sample
reviews_sample = sample['number of reviews'].values

# Define a function to calculate the median
def median_fn(data):
    return np.median(data)

# Perform bootstrapping
n_iterations = 1000
bootstrap_medians = []

for _ in range(n_iterations):
    sample_indices = resample(np.arange(len(reviews_sample)), replace=True, n_samples=len(reviews_sample))
    bootstrap_sample = reviews_sample[sample_indices]
    bootstrap_medians.append(median_fn(bootstrap_sample))

# Calculate the 95% confidence interval using the percentile method
confidence_interval = np.percentile(bootstrap_medians, [2.5, 97.5])

# Extract the median of the sample
median_reviews = np.median(df['number of reviews'].dropna())

# Plot the histogram and mark the median and confidence interval
plt.figure(figsize=(10, 6))
sns.histplot(reviews_sample, kde=False, color="blue", bins=10, alpha=0.5)
plt.axvline(x=median_reviews, color='red', linestyle='--', linewidth=1, label=f'Median: {median_reviews:.2f}')
plt.axvline(x=confidence_interval[0], color='green', linestyle=':', linewidth=1, label=f'95% CI Lower: {confidence_interval[0]:.2f}')
plt.axvline(x=confidence_interval[1], color='green', linestyle=':', linewidth=1, label=f'95% CI Upper: {confidence_interval[1]:.2f}')
plt.title('Histogram of Number of Reviews (Sample of 25)')
plt.xlabel('Number of Reviews')
plt.ylabel('Frequency')
plt.legend()
plt.show()

confidence_interval, median_reviews
```

## **B)**

**Pick a random sample of size 20. Then, using the bootstrapping
method, calculate a 95% confidence interval for the mean of this
variable using the standard error method. Also, plot the bootstrap
distribution on the dot-plot.**


```python
import numpy as np
import matplotlib.pyplot as plt

# Pick a random sample of size 20
sample = np.random.choice(df['number of reviews'], size=20, replace=False)

# Bootstrap the sample to calculate the 95% confidence interval for the mean
bootstrap_means = []
for _ in range(1000):  # number of bootstrap samples
    bootstrap_sample = np.random.choice(sample, size=20, replace=True)
    bootstrap_means.append(np.mean(bootstrap_sample))

# Calculate the 95% confidence interval using the standard error
bootstrap_mean = np.mean(bootstrap_means)
bootstrap_se = np.std(bootstrap_means)
ci_lower = bootstrap_mean - 1.96 * bootstrap_se
ci_upper = bootstrap_mean + 1.96 * bootstrap_se

# Plotting the bootstrap distribution and the confidence interval
plt.figure(figsize=(10, 5))
plt.hist(bootstrap_means, bins=30, alpha=0.7, label='Bootstrap Distribution')
plt.axvline(x=bootstrap_mean, color='r', linestyle='-', label='Mean of Bootstrap Means')
plt.axvline(x=ci_lower, color='g', linestyle='--', label='Lower 95% CI')
plt.axvline(x=ci_upper, color='b', linestyle='--', label='Upper 95% CI')
plt.xlabel('Mean Value')
plt.ylabel('Frequency')
plt.title('Bootstrap Distribution of the Mean with 95% Confidence Interval')
plt.legend()
plt.show()

# Output the results
print("Bootstrap Mean:", bootstrap_mean)
print("95% Confidence Interval:", (ci_lower, ci_upper))
```

The histogram depicts the bootstrap distribution of the mean from a sampled dataset, with a bootstrap mean marked at $27.95$ and a 95% confidence interval ranging from $16.33$ to $39.57$, indicating substantial variability in the mean estimates. This wide confidence interval reflects significant uncertainty in the mean estimation, suggesting the presence of considerable sample variability or a small sample size.
